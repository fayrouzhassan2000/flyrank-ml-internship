## 2. Data

This project uses the **FlyRank ML Internship warehouse release** hosted on Hugging Face.

The analysis is built from FlyRank content-performance data, including the `fact_content_daily_performance` table and the `dim_content` table. These sources provide historical search visibility, clicks, average position, engagement, and content-level performance signals.

For the modeling dataset, the analysis uses the following public-safe fields:

* `client_hash_id`
* `content_hash_id`
* `gsc_impressions`
* `gsc_clicks`
* `gsc_avg_position`
* `ga4_sessions`
* `content_type`
* `april_impressions`
* `is_declining_label`

The modeling data combines historical performance signals with a future-period impression measure (`april_impressions`) to identify whether content experienced a decline, represented by `is_declining_label`.

The analysis excludes client names, domains, URLs, private search queries, credentials, raw exports, and other identifying information. Only hashed identifiers and aggregated performance signals are used in the public-facing research.
